## Paso 1. Importación de librerías

````python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 50)
pd.set_option('display.float_format', '{:,.2f}'.format)
````

## Paso 2. Carga del CSV

````python
df_raw = pd.read_csv(
    'encuesta_snacks_mundial_2026_guatemala_2500_respuestas.csv'
)

print('Dimensiones del dataset:', df_raw.shape)
df_raw.head(3)
````

````python
print(df_raw.columns.tolist())
````

## Paso 3. Selección de columnas — todas las secciones

Se extraen todas las columnas a utilizar

````python
todas_las_columnas = [
    'EncuestaID',
    'FechaEncuesta',

    # Sección 1
    'RangoEdad',
    'Genero',
    'Departamento',
    'Municipio',
    'Ocupacion',

    # Sección 2
    'FrecuenciaConsumoSnacks',
    'LugarCompraSnacks',
    'ConQuienVePartidos',
    'GastoSnacksPartido',

    # Sección 3
    'SnacksSeleccionados',
    'SaborPreferido',
    'PresentacionPreferida',

    # Sección 4
    'PrecioAdecuado',
    'PagaMasEdicionMundial',

    # Sección 5
    'PlaneaVerMundial2026',
    'SeleccionApoya',
    'CompraDisenoSeleccion',
    'SeleccionInfluyeCompra',

    # Sección 6
    'JugadoresInfluyentes',
    'TipoPublicidadAtractiva',

    # Sección 7
    'PromocionPreferida',
    'CompraTarjetasColeccionables',
    'CampaniaMasProbableCompra'
]

df = df_raw[todas_las_columnas].copy()

print('Dimensiones del dataset unificado:', df.shape)
df.head()
````

````python
print('Registros duplicados:', df.duplicated().sum())
print('EncuestaID duplicados:', df['EncuestaID'].duplicated().sum())
````

## Paso 5. Limpieza de datos

````python
df = df.rename(columns={
    'PrecioAdecuado': 'PrecioIdealSnack',
    'PagaMasEdicionMundial': 'PagariaEdicionEspecial'
})
````

### 5.1 Corrección de valores nulos

````python
# Secciones 1-3
df['Ocupacion']          = df['Ocupacion'].fillna('No Indicado')
df['LugarCompraSnacks']  = df['LugarCompraSnacks'].fillna('No Indicado')
df['SaborPreferido']     = df['SaborPreferido'].fillna('No Indicado')

# Secciones 4-7
df['PrecioIdealSnack']          = df['PrecioIdealSnack'].fillna('No Indicado')
df['TipoPublicidadAtractiva']   = df['TipoPublicidadAtractiva'].fillna('No Indicado')
df['PromocionPreferida']        = df['PromocionPreferida'].fillna('No Indicado')
df['CampaniaMasProbableCompra'] = df['CampaniaMasProbableCompra'].fillna('No Indicado')
df['SeleccionApoya']            = df['SeleccionApoya'].fillna('No Indicado')
df['SeleccionInfluyeCompra']    = df['SeleccionInfluyeCompra'].fillna('No Indicado')

print('Nulos restantes:', df.isnull().sum().sum())
````

````python
df.isnull().sum()[df.isnull().sum() > 0]
````

````python
df['Municipio'] = df['Municipio'].fillna('No Indicado')
````

````python
print(df.isnull().sum().sum())
````

### 5.2 Normalización de texto

````python
columnas_texto = [
    'RangoEdad', 'Genero', 'Departamento', 'Municipio', 'Ocupacion',
    'FrecuenciaConsumoSnacks', 'LugarCompraSnacks', 'ConQuienVePartidos',
    'GastoSnacksPartido', 'SaborPreferido', 'PresentacionPreferida',
    'PrecioIdealSnack', 'PagariaEdicionEspecial', 'PlaneaVerMundial2026',
    'SeleccionApoya', 'CompraDisenoSeleccion', 'SeleccionInfluyeCompra',
    'TipoPublicidadAtractiva', 'PromocionPreferida',
    'CompraTarjetasColeccionables', 'CampaniaMasProbableCompra'
]

for col in columnas_texto:
    df[col] = df[col].astype(str).str.strip().str.title()

# Campos multi-valor: normalizar separador
for campo in ['SnacksSeleccionados', 'JugadoresInfluyentes']:
    df[campo] = (
        df[campo].astype(str)
        .apply(lambda x: '; '.join([v.strip() for v in x.split(';')]))
    )

print('Normalización completada.')
print('\nValores únicos de SaborPreferido:')
print(sorted(df['SaborPreferido'].unique()))
````

### 5.3 Conversión de tipos de datos

````python
df['FechaEncuesta'] = pd.to_datetime(df['FechaEncuesta'])

meses_es = {
    1: 'Enero', 2: 'Febrero', 3: 'Marzo', 4: 'Abril',
    5: 'Mayo', 6: 'Junio', 7: 'Julio', 8: 'Agosto',
    9: 'Septiembre', 10: 'Octubre', 11: 'Noviembre', 12: 'Diciembre'
}
df['FechaEncuesta'] = df['FechaEncuesta'].apply(
    lambda f: f'{f.day:02d} {meses_es[f.month]} {f.year}'
)
df['EncuestaID'] = df['EncuestaID'].astype(int)

print('Conversión de tipos completada.')
````

### 5.4 Órdenes lógicos para columnas ordinales

````python
orden_edad = [
    'Menos De 18 Años', '18 - 24 Años', '25 - 34 Años',
    '35 - 44 Años', '45 - 54 Años', '55 Años O Mas'
]

orden_gasto = [
    'Menos De Q25', 'Q25 - Q50', 'Q51 - Q100', 'Q101 - Q150', 'Más De Q150'
]

orden_frecuencia = [
    'Rara Vez', 'Varias Veces Al Mes', '1 A 2 Veces Por Semana',
    '3 A 5 Veces Por Semana', 'Todos Los Dias'
]

orden_precio = [
    'Menos De Q10', 'Q10 - Q20', 'Q21 - Q35', 'Q36 - Q50', 'Más De Q50'
]

print('Órdenes definidos correctamente.')
````

## Paso 6. Transformación de datos

### 6.1 Gasto promedio estimado en snacks por partido

````python
valores_gasto = df['GastoSnacksPartido'].unique().tolist()

mapa_gasto = {
    'Menos De Q25': 12.5,
    'Q25 - Q50':    37.5,
    'Q51 - Q100':   75.5,
    'Q101 - Q150': 125.5
}
for v in valores_gasto:
    if v not in mapa_gasto:
        mapa_gasto[v] = 175.0

df['GastoPromedio'] = df['GastoSnacksPartido'].map(mapa_gasto)

print('Cobertura del mapa de gasto:', df['GastoPromedio'].isnull().sum(), 'sin mapear')
print(f'Gasto promedio estimado global: Q{df["GastoPromedio"].mean():,.2f}')
````

### 6.2 Precio promedio ideal estimado por rango

````python
valores_precio = df['PrecioIdealSnack'].unique().tolist()

mapa_precio = {
    'Menos De Q10': 7.5,
    'Q10 - Q20':   15.0,
    'Q21 - Q35':   28.0,
    'Q36 - Q50':   43.0,
}
for v in valores_precio:
    if v not in mapa_precio:
        mapa_precio[v] = 57.5

df['PrecioPromedio'] = df['PrecioIdealSnack'].map(mapa_precio)

print('Cobertura del mapa de precio:', df['PrecioPromedio'].isnull().sum(), 'sin mapear')
print(f'Precio promedio ideal estimado global: Q{df["PrecioPromedio"].mean():,.2f}')
````

### 6.3 Flag de alta intención de compra ligada al Mundial

````python
df['AltaIntencionCompra'] = (
    (df['PlaneaVerMundial2026'].str.lower().str.contains('sí|si', na=False)) &
    (df['CompraDisenoSeleccion'].str.lower().str.contains('sí|si|probablemente|definitivamente', na=False))
).astype(int)

print(f'Encuestados con alta intención de compra: {df["AltaIntencionCompra"].sum():,}')
print(f'Porcentaje del total: {df["AltaIntencionCompra"].mean()*100:.2f}%')
````

## Paso 7. Construcción del Star Schema

### 7.1 dim_encuestado

````python
dim_encuestado = df[[
    'EncuestaID', 'RangoEdad', 'Genero', 'Departamento', 'Municipio', 'Ocupacion'
]].copy()

print('dim_encuestado:', dim_encuestado.shape)
dim_encuestado.head()
````

### 7.2 dim_producto

````python
dim_producto = df[[
    'EncuestaID', 'SaborPreferido', 'PresentacionPreferida'
]].copy()

print('dim_producto:', dim_producto.shape)
dim_producto.head()
````

### 7.3 dim_snacks_detalle

Se desanida `SnacksSeleccionados` (separado por `;`) para obtener una fila por snack,
lo que permite contar menciones individuales correctamente.

````python
dim_snacks_detalle = (
    df[['EncuestaID', 'SnacksSeleccionados']].copy()
    .assign(Snack=lambda d: d['SnacksSeleccionados'].str.split('; '))
    .explode('Snack')
    [['EncuestaID', 'Snack']]
    .assign(Snack=lambda d: d['Snack'].str.strip())
    .reset_index(drop=True)
)

print('dim_snacks_detalle:', dim_snacks_detalle.shape)
print('Snacks únicos:', dim_snacks_detalle['Snack'].nunique())
dim_snacks_detalle.head(8)
````

### 7.4 dim_precio

````python
dim_precio = df[[
    'EncuestaID', 'PrecioIdealSnack', 'PrecioPromedio', 'PagariaEdicionEspecial'
]].copy()

print('dim_precio:', dim_precio.shape)
dim_precio.head()
````

### 7.5 dim_mundial

````python
dim_mundial = df[[
    'EncuestaID', 'PlaneaVerMundial2026', 'SeleccionApoya',
    'CompraDisenoSeleccion', 'SeleccionInfluyeCompra', 'AltaIntencionCompra'
]].copy()

print('dim_mundial:', dim_mundial.shape)
dim_mundial.head()
````

### 7.6 dim_jugadores_detalle

Se desanida `JugadoresInfluyentes` (separado por `;`) para obtener una fila por jugador.

````python
dim_jugadores_detalle = (
    df[['EncuestaID', 'JugadoresInfluyentes']].copy()
    .assign(Jugador=lambda d: d['JugadoresInfluyentes'].str.split('; '))
    .explode('Jugador')
    [['EncuestaID', 'Jugador']]
    .assign(Jugador=lambda d: d['Jugador'].str.strip())
    .reset_index(drop=True)
)

# Excluir vacíos y "Ninguno"
dim_jugadores_detalle = dim_jugadores_detalle[
    ~dim_jugadores_detalle['Jugador'].str.lower().isin(['ninguno', 'nan', ''])
].reset_index(drop=True)

print('dim_jugadores_detalle:', dim_jugadores_detalle.shape)
print('Jugadores únicos:', dim_jugadores_detalle['Jugador'].nunique())
dim_jugadores_detalle.head(8)
````

### 7.7 dim_promociones

````python
dim_promociones = df[[
    'EncuestaID', 'TipoPublicidadAtractiva', 'PromocionPreferida',
    'CompraTarjetasColeccionables', 'CampaniaMasProbableCompra'
]].copy()

print('dim_promociones:', dim_promociones.shape)
dim_promociones.head()
````

### 7.8 fact_encuesta (tabla de hechos unificada)

````python
fact_encuesta = df[[
    'EncuestaID', 'FechaEncuesta',
    'FrecuenciaConsumoSnacks', 'LugarCompraSnacks',
    'ConQuienVePartidos', 'GastoSnacksPartido', 'GastoPromedio',
    'PrecioIdealSnack', 'PrecioPromedio', 'PagariaEdicionEspecial',
    'PlaneaVerMundial2026', 'CompraDisenoSeleccion', 'AltaIntencionCompra'
]].copy()

print('fact_encuesta:', fact_encuesta.shape)
fact_encuesta.head()
````

## Paso 8. KPIs ejecutivos

````python
total_encuestados        = len(df)
total_departamentos      = df['Departamento'].nunique()
total_municipios         = df['Municipio'].nunique()
gasto_promedio_global    = df['GastoPromedio'].mean()
precio_promedio_global   = df['PrecioPromedio'].mean()
total_menciones_snacks   = len(dim_snacks_detalle)
total_menciones_jugadores= len(dim_jugadores_detalle)
pct_ve_mundial           = (
    df['PlaneaVerMundial2026'].str.lower().str.contains('sí|si', na=False).sum()
    / total_encuestados * 100
)
pct_alta_intencion       = df['AltaIntencionCompra'].mean() * 100

print('════════════════════════════════════════════')
print('  KPIs EJECUTIVOS — ENCUESTA COMPLETA')
print('════════════════════════════════════════════')
print(f'  Total de encuestados:            {total_encuestados:,}')
print(f'  Departamentos cubiertos:         {total_departamentos}')
print(f'  Municipios cubiertos:            {total_municipios}')
print(f'  Gasto promedio por partido:      Q{gasto_promedio_global:,.2f}')
print(f'  Precio ideal promedio estimado:  Q{precio_promedio_global:,.2f}')
print(f'  Total menciones de snacks:       {total_menciones_snacks:,}')
print(f'  Total menciones de jugadores:    {total_menciones_jugadores:,}')
print(f'  Planea ver el Mundial:           {pct_ve_mundial:.1f}%')
print(f'  Alta intención de compra:        {pct_alta_intencion:.1f}%')
````

## Paso 9. Reportes para directivos

### 9.1 Perfil demográfico: distribución por departamento y ocupación

````python
perfil_depto_ocupacion = (
    df.groupby(['Departamento', 'Ocupacion'])
    .agg(Encuestados=('EncuestaID', 'count'))
    .reset_index()
    .sort_values(['Departamento', 'Encuestados'], ascending=[True, False])
)
perfil_depto_ocupacion['Porcentaje'] = (
    perfil_depto_ocupacion['Encuestados'] / total_encuestados * 100
).round(2)

print('Perfil del consumidor por departamento y ocupación:')
perfil_depto_ocupacion.head(20)
````

> **Interpretación:** La tabla muestra la concentración geográfica y ocupacional de la muestra.
> Los departamentos con mayor volumen de respuestas son los mercados prioritarios para la distribución
> de snacks temáticos. La ocupación dominante en cada departamento orienta el tono y canal
> de comunicación a estudiantes y empleados privados responden mejor a promociones digitales y 2x1,
> mientras que profesionales y empresarios prefieren ediciones premium.

### 9.2 Género: distribución global y por perfil demográfico

````python
genero_global = (
    df['Genero'].value_counts(normalize=True)
    .mul(100).round(2).reset_index()
)
genero_global.columns = ['Genero', 'Porcentaje']

print('Género — Porcentaje global:')
print(genero_global.to_string(index=False))
````

````python
genero_edad_ocupacion = (
    df.groupby(['RangoEdad', 'Ocupacion', 'Genero'])
    .agg(Encuestados=('EncuestaID', 'count'))
    .reset_index()
    .sort_values(['RangoEdad', 'Ocupacion', 'Encuestados'], ascending=[True, True, False])
)
genero_edad_ocupacion['Porcentaje'] = (
    genero_edad_ocupacion['Encuestados'] / total_encuestados * 100
).round(2)

print('Género por rango de edad y ocupación:')
genero_edad_ocupacion.head(20)
````

> **Interpretación:** La distribución de género permite segmentar creatividades publicitarias.
> Si el género masculino domina en rangos 25-44, las campañas centradas en jugadores estrella
> (Messi, Mbappé) y ediciones de selecciones tienen mayor retorno en ese segmento.
> Para segmentos con mayor presencia femenina, las propuestas de snacks de sabores dulces
> (Caramel Goal Popcorn, ChocoKick Bites) y presentaciones individuales pueden tener mejorvista para ellas.

### 9.3 Frecuencia de consumo de snacks

````python
frecuencia_rank = (
    df['FrecuenciaConsumoSnacks'].value_counts().reset_index()
)
frecuencia_rank.columns = ['FrecuenciaConsumoSnacks', 'Encuestados']
frecuencia_rank['Porcentaje'] = (
    frecuencia_rank['Encuestados'] / total_encuestados * 100
).round(2)
frecuencia_rank.insert(0, 'Posicion', range(1, len(frecuencia_rank) + 1))

print('Ranking de frecuencia de consumo (global):')
print(frecuencia_rank.to_string(index=False))
````

````python
frecuencia_edad_ocupacion = (
    df.groupby(['RangoEdad', 'Ocupacion', 'FrecuenciaConsumoSnacks'])
    .agg(Encuestados=('EncuestaID', 'count'))
    .reset_index()
    .sort_values(['RangoEdad', 'Ocupacion', 'Encuestados'], ascending=[True, True, False])
)
frecuencia_edad_ocupacion['Porcentaje'] = (
    frecuencia_edad_ocupacion['Encuestados'] / total_encuestados * 100
).round(2)

print('Frecuencia de consumo por rango de edad y ocupación:')
frecuencia_edad_ocupacion.head(20)
````

> **Interpretación:** La frecuencia de consumo define la intensidad de la oportunidad de negocio.
> Consumidores que compran snacks 3-5 veces por semana o diariamente son el centro del volumen de ventas;
> activarlos con tarjetas coleccionables del Mundial o empaques de jugadores como Messi o Mbappé
> genera compras recurrentes. Los consumidores ocasionales ("Rara Vez") pueden ser manejados
> con ediciones especiales de selecciones y promociones de tiempo limitado durante los partidos.

````python
top3_lugar = lugar_rank.head(3)[['Posicion', 'LugarCompraSnacks', 'Porcentaje']]
print('Top 3 canales de compra:')
print(top3_lugar.to_string(index=False))
````

### 9.5 Compañía al ver los partidos

````python
compania_rank = (
    df['ConQuienVePartidos'].value_counts().reset_index()
)
compania_rank.columns = ['ConQuienVePartidos', 'Encuestados']
compania_rank['Porcentaje'] = (
    compania_rank['Encuestados'] / total_encuestados * 100
).round(2)
compania_rank.insert(0, 'Posicion', range(1, len(compania_rank) + 1))

print('Ranking — con quién ve los partidos (global):')
print(compania_rank.to_string(index=False))
````

### 9.6 Presupuesto destinado a snacks por partido

````python
gasto_rank = (
    df['GastoSnacksPartido'].value_counts().reset_index()
)
gasto_rank.columns = ['GastoSnacksPartido', 'Encuestados']
gasto_rank['Porcentaje'] = (
    gasto_rank['Encuestados'] / total_encuestados * 100
).round(2)
gasto_rank.insert(0, 'Posicion', range(1, len(gasto_rank) + 1))

print('Ranking de presupuesto por partido (global):')
print(gasto_rank.to_string(index=False))
print()
print(f'Gasto promedio estimado global: Q{df["GastoPromedio"].mean():,.2f}')
````

> en ese rango para maximizar conversión. Los segmentos de mayor gasto (Q101+) son los candidatos
> ideales para ediciones premium coleccionables con diseño de selecciones o jugadores estrella.

### 9.7 Snacks favoritos: top 10 y top 3

> del consumidor guatemalteco. Si los fans de Brasil mencionan masivamente a Vinícius Jr.,
> ese es el duo para el empaque de esa edición. Si los fans de Argentina se inclinan
> por Messi y Julián Álvarez, ambos pueden aparecer en la misma variante coleccionable.
> Las celdas con mayor valor son las combinaciones embajador-selección de mayor impacto comercial
> y deben ser la guía para contratar derechos de imagen y diseñar los empaques coleccionables.

Para diseñar la campaña de snacks orientada al Mundial FIFA 2026, realizamos una encuesta con 2,500 respuestas a nivel nacional, abarcando múltiples departamentos y municipios de Guatemala. El objetivo fue entender los hábitos de consumo, las preferencias de producto, la afinidad con el fútbol y la disposición de compra del consumidor guatemalteco, para que cada decisión de la campaña esté respaldada por datos reales.

**¿Quién es nuestro consumidor?**

El perfil del encuestado varía entre estudiantes y empleados privados como grupos dominantes, distribuidos principalmente en los departamentos con mayor densidad poblacional. Esta concentración geográfica nos dice dónde enfocar la distribución del producto temático y qué tono de comunicación usar: los estudiantes y jóvenes responden mejor a promociones digitales y mecánicas tipo 2x1, mientras que profesionales y personas de mayor poder adquisitivo son candidatos naturales para ediciones premium coleccionables.

**Consumo y gasto**

La frecuencia de consumo de snacks es alta en una buena parte de la muestra, con consumidores habituales que compran varias veces por semana o incluso a diario. Estos consumidores frecuentes son el corazón de la campaña: activarlos con empaques temáticos del Mundial o tarjetas coleccionables de jugadores genera compras recurrentes. Los consumidores ocasionales, en cambio, pueden capturarse con ediciones especiales de tiempo limitado durante los partidos clave.
En cuanto al gasto por partido, el rango predominante se sitúa entre Q25 y Q50, lo que nos da el precio óptimo de posicionamiento para los snacks temáticos. Los segmentos que gastan más de Q100 son los ideales para una línea premium coleccionable con un sobreprecio de 20 a 30% respecto al precio base estándar.

**Producto:** 

qué snack, qué sabor y en qué presentación
El análisis del top 10 de snacks favoritos nos da el portafolio mínimo a activar con la campaña FIFA 2026. El snack número uno del ranking es el producto ancla: su empaque debe ser el primero en recibir el diseño temático. El top 3 concentra la mayoría de las menciones y representa los tres productos con mayor retorno si se les da visibilidad temática.

**El sabor líder define la prioridad de producción:** 

si el picante o el BBQ encabeza el ranking, esos sabores deben adoptar primero el empaque del Mundial. En cuanto a la presentación, el formato más votado —ya sea bolsón mediano, individual o multipack— es el SKU prioritario para el rediseño. Las presentaciones familiares tienen sentido para quienes ven los partidos en grupo, que según nuestra encuesta representan la mayoría de los consumidores.

**¿Qué selección va en el empaque?**

La selección con más apoyo entre los guatemalteco es la que debe protagonizar el diseño principal. El top 3 de selecciones apoyadas define las tres variantes de empaque coleccionable a producir. Sin embargo, el criterio definitivo no es solo la selección más querida, sino cuál tiene mayor poder de conversión en ventas, lo que puede diferir: un consumidor puede apoyar a Brasil pero ser motivado a comprar si ve a Messi en el empaque. Por eso cruzamos la selección apoyada con la selección que más influye en la decisión de compra para tomar la mejor decisión.

**El jugador embajador**

El jugador número uno del ranking de influencia es el embajador natural de la campaña. Si Messi encabeza, su imagen en el snack más vendido con el diseño de Argentina genera la combinación de mayor impacto en ventas. El cruce por edad nos permite saber qué jugador resuena con cada generación: jugadores como Vinícius Jr. o Lamine Yamal pueden ser más efectivos en el segmento de 18 a 24 años, mientras Messi y Ronaldo dominan en el rango de 25 a 44 años.

**Publicidad y promociones**

El tipo de publicidad más atractiva para el consumidor guatemalteco define dónde invertir el presupuesto de medios. Si el contenido con jugadores famosos lidera, el creativo principal debe girar en torno a la figura embajadora identificada. En segmentos jóvenes, el contenido en redes sociales y videos cortos supera a los formatos tradicionales.

En cuanto a promociones, el 2x1 durante los partidos de mayor audiencia —fases eliminatorias y final— maximiza el volumen de ventas en los momentos pico. Las tarjetas coleccionables, si superan el 55% de interés en nuestra encuesta, están respaldadas por datos para incluirse como mecánica principal, especialmente en el segmento menor de 24 años.

**Conclusión**

La campaña de snacks para el Mundial FIFA 2026 tiene una base sólida en los datos. El producto ancla es el snack líder del ranking, en el sabor dominante, en la presentación más votada, con el diseño de la selección de mayor influencia en compra y la imagen del jugador más mencionado. El precio debe ubicarse en el rango que concentra la mayor disposición de pago, con una variante premium para los segmentos de mayor gasto. Las promociones tipo 2x1 y tarjetas coleccionables son las mecánicas de mayor conversión, y los canales de activación prioritarios son los supermercados y tiendas de conveniencia donde hoy se concentra la compra habitual de snacks.
